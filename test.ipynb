```python
import sympy as sp

x = sp.symbols('x')
root = (8*x**3 + 8*x)**(sp.Rational(1, 4))
F = 1/(4*2**(sp.Rational(1, 4))) * (sp.log((x + 1 - root)/(x + 1 + root)) + 2*sp.atan(root/(x + 1)))

# To see it simplified
sp.simplify(F)
```

```python
integrand = 1/((x - 1)*(x**3 + x)**(sp.Rational(1, 4)))
F_diff = sp.diff(F, x)
```

```python
F_diff
```

```python
F_diff.simplify()
```

```python
F_diff.equals(integrand)
```
